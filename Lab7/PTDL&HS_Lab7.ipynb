```python
import random
reviews = nltk.corpus.movie_reviews
documents = [(list(reviews.words(fileid)), category)
for category in reviews.categories()
for fileid in reviews.fileids(category)]
random.shuffle(documents)
```

```python
first_review = ' '.join(documents[0][0])
print(first_review)
```

```python
documents[0][1]
```

```python
all_words = nltk.FreqDist(w.lower() for w in reviews.words())
word_features = list(all_words)
```

```python
def document_features(document, word_features):
    document_words = set(document)
    features = {}
    for word in word_features:
        features['{}'.format(word)] = (word in document_words)
    return features
```

```python
featuresets = [(document_features(d,word_features), c) for (d,c) in documents]
len(featuresets)
```

```python
train_set, test_set = featuresets[1500:], featuresets[:500]
classifier = nltk.NaiveBayesClassifier.train(train_set)
```

```python
train_set, est_set = featuresets[1500:], featuresets[:500]
classifier = nltk.NaiveBayesClassifier.train(train_set)
print(nltk.classify.accuracy(classifier, test_set))
```

7. Viết chương trình Python với thư viện NLTK để tìm tập hợp các từ đồng nghĩa và trái nghĩa
của một từ nào đó.


```python
tu_khoa_7 = "good"
dong_nghia = set()
trai_nghia = set()

for syn in wn.synsets(tu_khoa_7):
    for lemma in syn.lemmas():
        dong_nghia.add(lemma.name())
        # Nếu có từ trái nghĩa
        if lemma.antonyms():
            trai_nghia.add(lemma.antonyms()[0].name())

print(f"Từ khóa: {tu_khoa_7}")
print("Từ đồng nghĩa:", list(dong_nghia)[:10]) # In tối đa 10 từ
print("Từ trái nghĩa:", list(trai_nghia)[:10])
```

9. Viết chương trình Python với thư viện NLTK để so sánh sự giống nhau của hai danh từ đã
cho.


```python
danh_tu_1 = wn.synset('cat.n.01')
danh_tu_2 = wn.synset('dog.n.01')
danh_tu_3 = wn.synset('car.n.01')
do_tuong_dong_1_2 = danh_tu_1.wup_similarity(danh_tu_2) # Wu-Palmer Similarity
do_tuong_dong_1_3 = danh_tu_1.wup_similarity(danh_tu_3)

print(f"Sự giống nhau giữa Mèo (cat) và Chó (dog): {do_tuong_dong_1_2:.4f}")
print(f"Sự giống nhau giữa Mèo (cat) và Ô tô (car): {do_tuong_dong_1_3:.4f}")
```

10. Viết chương trình Python với thư viện NLTK để so sánh sự giống nhau của hai động từ đã
cho.


```python
dong_tu_1 = wn.synset('run.v.01')
dong_tu_2 = wn.synset('walk.v.01')
dong_tu_3 = wn.synset('eat.v.01')

tuong_dong_v_1_2 = dong_tu_1.wup_similarity(dong_tu_2)
tuong_dong_v_1_3 = dong_tu_1.wup_similarity(dong_tu_3)

print(f"Sự giống nhau giữa Chạy (run) và Đi bộ (walk): {tuong_dong_v_1_2:.4f}")
print(f"Sự giống nhau giữa Chạy (run) và Ăn (eat): {tuong_dong_v_1_3:.4f}")
```

11. Viết chương trình Python với thư viện NLTK để tìm số lượng tên nam và nữ trong các tên
kho ngữ liệu. In tên 10 nam và nữ đầu tiên. Lưu ý: Kho văn bản tên chứa tổng cộng khoảng
2943 nam (male.txt) và 5001 nữ (Female.txt) tên. Kho được biên soạn bởi Kantrowitz, Ross.


```python
ten_nam = names.words('male.txt')
ten_nu = names.words('female.txt')

print(f"Tổng số tên nam: {len(ten_nam)}")
print(f"Tổng số tên nữ: {len(ten_nu)}")
print("10 tên nam đầu tiên:", ten_nam[:10])
print("10 tên nữ đầu tiên:", ten_nu[:10])
```

12. Viết chương trình Python với thư viện NLTK để in 15 kết hợp ngẫu nhiên đầu tiên được gắn
nhãn nam và được gắn nhãn tên nữ từ kho tên.


```python
danh_sach_nhan = [(ten, 'male') for ten in ten_nam] + [(ten, 'female') for ten in ten_nu]

random.seed(42)
random.shuffle(danh_sach_nhan)
for item in danh_sach_nhan[:15]:
    print(item)
```
